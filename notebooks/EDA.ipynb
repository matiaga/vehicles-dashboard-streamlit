## 2. Preparación de los datos <a id='data_preprocessing'></a>

#### Conviertimos los datos <a id='p2_conversion'></a>

Primero vamos a revisar los datos de las columnas `model_year`,`is_4wd` y `cylinders`.

```python
# Revisemos los valores únicos en la columna `model_year`. 
car_data['model_year'].unique()
```

```python
# Revisemos los valores únicos en la columna `is_4wd`. 
car_data['is_4wd'].unique()
```

```python
# Revisemos los valores únicos en la columna `cylinders`. 
car_data['cylinders'].unique()
```

* Las columnas `model_year`, `cylinders`, y `is_4wd` las transformaremos a enteros `int64`.
* La coluumna `is_4wd` representa si un vehiculo tiene tracción en las 4 llantas o no, por lo que identificaremos los registros como `4WD` en caso de 1 o `2WD` en caso de 0 en una columna adicional `traccion`.
* La columna `date_poste` la transformamos a tipo `datetime`.

```python
# Cambios de tipos de datos
car_data["date_posted"] = pd.to_datetime(car_data["date_posted"])
car_data["model_year"] = car_data["model_year"].astype("Int64")
car_data["cylinders"] = car_data["cylinders"].astype("Int64")
car_data["is_4wd"] = car_data["is_4wd"].astype("Int64").fillna(0)

# Creación columna traccion
# Tracción de 4 llantas o de dos. 4WD: Four Wheel Drive
car_data["traccion"] = car_data["is_4wd"].map({1: "4WD", 0: "2WD"})
```

```python
car_data.head()
```

#### Revisamos los valores ausentes <a id='p2_val_ausentes'></a>

```python
# 3. Revisión y manejo de valores ausentes
missing_val = car_data.isnull().sum().to_frame("Null values")

# Calcula el % de valores ausentes
missing_val["% of missing"] = round(missing_val['Null values'] / len(car_data) * 100,2)

missing_val = missing_val[missing_val['Null values'] > 0]

# Agregamos una columna con la acción a tomar con cada columna
missing_val['Accion'] = ["Imputar", "Imputar", "Imputar", "Reemplazar por Unknown"]

missing_val 
```

En la columna `model_year`, perder el 7% de los datos solo por esta columna no compensa, por lo tanto vamos a imputar con la mediana en lugar de eliminar los registros, agrupando primero por modelo. Por ejemplo, un Toyota Corolla suele pertenecer a un rango de años similar.

En `cylinders` con más de 10% de valores ausentes vamos a imputar con la moda y por modelo, porque un modelo específico suele ofrecerse con un número limitado de motores.

La columna `odometer` es una variable importante para elaborar el dashboard, por no lo que no vamos a eliminar el 15% de la data. También vamos a imputar con la mediana por el año del modelo `model_year`. El año de fabricación normalmente va a estar relacionado con el kilometraje.

En `paint_color` solo vamos a agregar una categoría `Unknown`. 

```python
# Imputar los valores nullos con la mediana agrupando por modelo
car_data["model_year"] = (
    car_data
    .groupby("model")["model_year"]
    .transform(lambda x: x.fillna(x.median().round()))
    #.astype("Int64")
)

# Imputar con la moda agrupando por modelo
car_data["cylinders"] = (
    car_data
    .groupby("model")["cylinders"]
    .transform(lambda x: x.fillna(x.mode().iloc[0]))
)

# Imputar con la mediana y por año de modelo
car_data["odometer"] = (
    car_data
    .groupby("model_year")["odometer"]
    .transform(lambda x: x.fillna(x.median()))
)

# Asignamos el color Unknown para los valores nulos
car_data["paint_color"] = (
    car_data["paint_color"]
    .fillna("Unknown")
)
```

```python
# Revisamos si aún hay valores ausentes
car_data.isnull().sum()
```

```python
# Revisamos el dato ausente en odometer
car_data[car_data['odometer'].isnull()]
```

```python
# Busquemos los vehículos del año 1929 que aún tiene un nulo.
car_data[ (car_data['model_year']==1929)]
```

Como solo hay un registro en el año 1929 no hay media. Por lo tanto, eliminaremos este único registro. 

```python
# Eliminamos los valores nulos de odometer restantes (1)
car_data.dropna(subset=["odometer"], inplace=True)
print(f"Hay {len(car_data)} registros en vehicles_us.csv y {car_data.shape[1]} columnas.\n" )
```

## 3. Análisis  <a id='analisis'></a>

### Distribución de precios <a id='p3_dist_precio'></a>

El precio es una de las variables principales para entender el comportamiento del mercado de vehículos usados, por ello, analizaremos la distribución de precios.
Buscamos identificar el rango de precios más común entre los vehículos publicados, así como posibles valores extremos.

```python
# Crear el histograma utilizando plotly.graph_objects
# Se crea una figura vacía y luego se añade un rastro de histograma
fig_price = go.Figure(data=[go.Histogram(x=car_data['price'], nbinsx=100)])

# Personalizamos y estilizamos el gráfico.
fig_price.update_layout(
    title={
        "text": "Distribución de precios",
        "x": 0.3
    },
    xaxis_title="Precio (USD)",
    yaxis_title="Cantidad de vehículos",
    template="plotly_white",
)

# Mostrar el gráfico Plotly
fig_price.show()
```

La distribución de precios permite observar que la mayoría de los vehículos se concentra en rangos de precio bajos y medios, hasta 50K, mientras que existen algunos vehículos con precios considerablemente más altos (outliers). Esto sugiere una distribución sesgada hacia la derecha, común en mercados donde hay pocos productos de alto valor comparados con una mayoría de opciones más accesibles.

### Distribución del kilometraje <a id='p3_dist_kilometraje'></a>

El kilometraje es una variable clave en el análisis de vehículos usados, ya que suele estar relacionado con el desgaste del vehículo y puede influir directamente en su precio. Analizar su distribución ayuda a entender qué tan usados están los vehículos publicados en el dataset.

```python
# Histograma de kilometraje
fig_hist_odometer = px.histogram(
    car_data,
    x="odometer",
    nbins=50,
    title="Distribución del kilometraje",
    labels={"odometer": "Kilometraje (millas)"},
)
fig_hist_odometer.update_yaxes(title_text="")
fig_hist_odometer.show()
```

La distribución del kilometraje permite identificar la concentración de vehículos según su nivel de uso, aquellos con kilometraje bajo, medio o alto. También permite observar posibles valores extremos, que podrían corresponder a vehículos con uso intensivo o registros poco comunes.

### Relación entre precio y kilometraje <a id='p3_scatter'></a>

Después de analizar las distribuciones individuales de precio y kilometraje, es útil observar la relación entre ambas variables. Este gráfico de dispersión permite identificar si los vehículos con mayor kilometraje tienden a tener precios más bajos, lo cual sería esperable en el mercado de vehículos usados.

```python
# Crear un scatter plot 
fig_scatter_1 = px.scatter(
    car_data,
    x="odometer",
    y="price",
    title="Precio vs. kilometraje",
    opacity=0.5,
    labels={"odometer": "Kilometraje (millas)", "price": "Precio (USD)"},
)
fig_scatter_1.update_layout(template="plotly_white", title_x=0.3, )
fig_scatter_1.show()
```

En términos generales, se esperaría que los vehículos con mayor kilometraje tengan precios más bajos, aunque la relación puede no ser perfectamente lineal debido a otros factores como el año del modelo, la marca, la condición, el tipo de vehículo y sus características específicas.

### Precio según la condición del vehículo <a id='p3_boxplot'></a>

La condición del vehículo es un factor que puede influir significativamente en su valor de mercado. El diagrama de cajas (boxplot) permite comparar la distribución de precios entre las diferentes categorías de condición, identificando diferencias en la mediana, la dispersión y la presencia de valores atípicos.

```python
fig_plot_cond = px.box(
    car_data,
    x="price",
    y="condition",
    orientation="h",
    title="Diagráma de Cajas: Precio según condición del Vehículo",
    labels={"condition": "Condición", "price": "Precio (USD)"}
)

fig_plot_cond.show()
```

Debido a la existencia de valores atípicos en la variable precio, se incluye a continuación una versión ajustada del gráfico para facilitar la interpretación de las cajas principales.

```python
# Versión del boxplot hasta el percentil 99

price_limit = car_data["price"].quantile(0.99)

fig_plot_cond = px.box(
    car_data,
    x="price",
    y="condition",
    orientation="h",
    points=False,
    title="Diagrama de cajas: Precio según condición del vehículo (hasta percentil 99)",
    labels={
        "condition": "Condición",
        "price": "Precio (USD)"
    }
)

fig_plot_cond.update_layout(
    template="plotly_white",
    title_x=0
)

fig_plot_cond.update_xaxes(
    range=[0, price_limit]
)

fig_plot_cond.show()
```

El boxplot muestra que la condición del vehículo tiene una influencia observable sobre su precio. En general, los vehículos clasificados en mejores condiciones presentan precios medianos más altos, mientras que las categorías de menor condición tienden a concentrar precios inferiores.

Asimismo, se observa una amplia dispersión de precios dentro de cada categoría, lo que indica que la condición no es el único factor que determina el valor de un vehículo. Variables como el año del modelo, el kilometraje, la marca y otras características también contribuyen significativamente al precio final.

### Precio Promedio por Año del Modelo <a id='p3_linea'></a>

```python
# Gráfico de relación Precio y Año del Modelo
avg_price_by_year = (
    car_data.groupby("model_year")["price"]
    .mean()
    .reset_index()
    .sort_values("model_year")
)

fig_line_model = px.line(
    avg_price_by_year,
    x="model_year",
    y="price",
    title="Precio Promedio por Año del Modelo",
    labels={"model_year": "Año del Modelo", "price": "Precio (USD)"},
)
fig_line_model.update_layout(template="plotly_white", title_x=0.3, )
fig_line_model.show()
```

El análisis muestra una tendencia general en la que los vehículos correspondientes a modelos más recientes presentan precios promedio más elevados. Este comportamiento es consistente con lo esperado en el mercado de vehículos usados, donde la depreciación suele aumentar con la antigüedad del vehículo.

No obstante, se observan fluctuaciones en algunos años, lo que sugiere que el precio promedio también está influenciado por otros factores presentes en el conjunto de datos, como el modelo, el kilometraje, la condición y el tipo de vehículo disponible para cada año del modelo.
